## Pandas series

```python
g7_pop = pd.Series([67.2, 65.3, 126.7, 60.7, 46.4, 64.6, 66.0])
g7_pop
```

```python
#series can also have a name which is useful for labeling data.
g7_pop.name = 'G7 Population in millions'
g7_pop
```

```python
g7_pop.dtype
```

```python
#you can also check values of the series using the values attribute.
g7_pop.values
```

```python
g7_pop
```

```python
# you can also get elements of the series using the index attribute.
g7_pop[2]
```

```python
g7_pop[0:4]
```

```python
# you can also set the index of the series using the index attribute.
g7_pop.index = [
    'canada',
    'france',
    'germany',
    'italy',
    'japan',
    'united kingdom',
    'united states'
]
g7_pop
```

```python
# you can also bypass the creation of the series and create it directly with ordered dictionary.
g7_pop = pd.Series({
    'canada' : 67.2,
    'france' : 65.3,
    'germany' : 126.7,
    'italy' : 60.7,
    'japan' : 46.4,
    'united kingdom' : 64.6,
    'united states' : 66.0
}, name = 'G7 Population in millions')
g7_pop
```

## Indexing

```python
# you can use the 'index' of the element you are looking for
g7_pop['germany']
```

```python
g7_pop
```

```python
# you can also get two elements of the series
g7_pop[['japan' , 'germany']]
```

## Pandas DataFrame


```python
# when creating a dataframe, you can pass a dictionary of series to the constructor. 
# The keys of the dictionary will be used as the column names 
# and the values will be used as the data for each column.
df = pd.DataFrame({
    'population': [67.2, 65.3, 126.7, 60.7, 46.4, 64.6, 66.0],
    'surface area': [9.985, 0.551, 0.357, 0.301, 0.378, 0.243, 9.834],
    'continent': [
        'South america', 
        'Asia',  
        'Europe', 
        'Europe',
        'Europe',
        'Europe',
        'North america'
    ]
})
df
```

```python
#lets assign the index of the dataframe to be the country names.
df.index = [
    'canada',
    'france',
    'germany',
    'italy',
    'japan',
    'united kingdom',
    'united states'
] 
df
```

```python
## when getting the columns of a dataframe, 
# you can use the column name as an attribute of the dataframe.
df.columns
```

```python
# when getting the index of a dataframe, you can use the index attribute of the dataframe.
df.index
```

```python
# when you want to get quick info about the dataframe, you can use the info() method of the dataframe.
df.info()
```

```python
# when getting size of the dataframe, you can use the shape attribute of the dataframe.
df.size
```

```python
# when getting the shape of the dataframe, you can use the shape attribute of the dataframe.
df.shape
```

```python
# if you want to get the summary of the dataframe, you can use the describe() method of the dataframe.
df.describe()
```

```python
# when getting the types of the columns in the dataframe, you can use the dtypes attribute of the dataframe.
df.dtypes
```

## Indexing, Selection and Slicing 

```python
df
```

```python
# when selecting individual rows of the dataframe, you can use the loc[] method of the dataframe.
df.loc['italy']
```

```python
# we use iloc[] method to select rows by their integer index.
df.iloc[-1]
```

```python
df.iloc[3]
```

```python
df
```

```python
# when you want to select a column of the dataframe, you use the dataframe name 
# followed by the column name in square brackets.
df['continent']
```

```python
# you can also get a row (-up to) and a column by using the loc[] method of the dataframe.
df.loc['france' : 'united kingdom', 'population']
```

```python
#same using iloc[] method to get the same result as above.
df.iloc[1 : 5, 0]
```

```python
# the same as above but if you want to get two coloumns, you can pass a list of 
# column names to the loc[] method.
df.loc['france' : 'united kingdom', ['population', 'surface area']]
```

```python
# this time using iloc[] method to get the same result as above.
df.iloc[1 : 5, [0, 1]]
```
