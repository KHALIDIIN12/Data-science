## Pandas series

```python
g7_pop = pd.Series([67.2, 65.3, 126.7, 60.7, 46.4, 64.6, 66.0])
g7_pop
```

```python
#series can also have a name which is useful for labeling data.
g7_pop.name = 'G7 Population in millions'
g7_pop
```

```python
g7_pop.dtype
```

```python
#you can also check values of the series using the values attribute.
g7_pop.values
```

```python
g7_pop
```

```python
# you can also get elements of the series using the index attribute.
g7_pop[2]
```

```python
g7_pop[0:4]
```

```python
# you can also set the index of the series using the index attribute.
g7_pop.index = [
    'canada',
    'france',
    'germany',
    'italy',
    'japan',
    'united kingdom',
    'united states'
]
g7_pop
```

```python
# you can also bypass the creation of the series and create it directly with ordered dictionary.
g7_pop = pd.Series({
    'canada' : 67.2,
    'france' : 65.3,
    'germany' : 126.7,
    'italy' : 60.7,
    'japan' : 46.4,
    'united kingdom' : 64.6,
    'united states' : 66.0
}, name = 'G7 Population in millions')
g7_pop
```

## Indexing

```python
# you can use the 'index' of the element you are looking for
g7_pop['germany']
```

```python
g7_pop
```

```python
# you can also get two elements of the series
g7_pop[['japan' , 'germany']]
```
