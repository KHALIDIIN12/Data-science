## Indexing and Slicing of Matrices

```python
d = np.array([
    [1, 2, 3, 4], #index 0
    [5, 6, 7, 8], #index 1
    [9, 10, 11, 12] #index 2 or -1
])
```

```python
d[-1] # this will print the last list [9, 10, 11, 12]
```

```python
d[1, 0] # the 1 will print the second row and 0 is the first number in the second row which is 5
```

```python
d[-1, 1]
```

```python
d.ndim
```

```python
d[1:3] # gets the second row up to but not including the 3rd row which we do not have
       # basically just like a for loop *up to but not including*
```

```python
d[:,:1] # prints all rows but only prints 1 column in each row
```

```python
d[1:, :2] # prints from row 2 but only prints two columns
```

```python
d[2] = np.array([13, 14, 15, 16]) # replaces last row with the new numbers
```

```python
print(d) # so when we print d you will see the changes
```
