```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame ({
    'SEX' : ['M', 'F', 'F', 'D', '?'],
    'Age' : [29, 30, 24, 290, 25]
})
df
```

## finding unique values

```python
# the first step to clean invalid values is to notice them, then identify them and finally 
#handle them appropriately(remove them, replace them) etc.in our dataframe
#we have 'sex' column which only takes in values like 'M' and 'F' but we have a 'D' and '?'
#we are gonna use the .unique() method
df['SEX'].unique()
```

```python
# or we can use .value_counts()
df['SEX'].value_counts()
```

```python
#clearly if you see values like 'D' or '?' are not appropriate identifiers in "SEX"
#let's say you have been told that 'D' was a typo and it should be replace with 'F'
#for that we use the .replace() function
df['SEX'].replace('D', 'F')
```

```python
df[df['Age'] > 100]
```

```python
df.loc[df['Age'] > 100, 'Age'] = df.loc[df['Age'] > 100, 'Age'] / 100
df
```
