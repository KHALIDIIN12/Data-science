## Pandas Operations With Missing Values

```python
s = pd.Series([1, 2, np.nan])
```

## Dropping Values

```python
s.dropna()
```

## Filling nulls with a arbitrary values

```python
s.fillna(0)
```

## filling nulls with close values

```python
# the method argument is used to fill null values with other values close to that null one
s.fillna(method = 'ffill')
```

```python
#now backwards fill
s.fillna(method = 'bfill')
```

## checking if there are NAs

```python
#if there are missing values, s.dropna() will have less elements than s.
s.dropna().count()
```

## Spliting columns

```python
df = pd.DataFrame({
    'Data': [
        '1987_M_US _1',
        '1990?_M_UK_1',
        '1992_F_US_2',
        '1970?_M_   IT_1',
        '1985_F_I  T_2'
]})
df
```

```python
df.info()
```

#### You know that the single columns represent the values "year, Sex, Country and number of children", but it's all been grouped in the same column and separated by an underscore. Pandas has a convenient method named split that we can use in these situations:



```python
df['Data'].str.split('_')
```

```python
df
```

```python
# and if you want to create a dataframe out of your newly sorted dataframe
df['Data'].str.split('_',expand = True)
```

```python
df = df['Data'].str.split('_', expand = True)
```

```python
df.columns = ["Year", "Sex", "Country", "CH"]
df
```

```python
# we see our year column contains row years that have '?' so we check it by using .contain()
df["Year"].str.contains('\?')
```
